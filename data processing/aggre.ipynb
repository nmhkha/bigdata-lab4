# Task 3 — Aggregation

```python
import pandas as pd
import matplotlib.pyplot as plt

df = pd.read_parquet("data_cleaned.parquet", engine="fastparquet")
df.head()
```

```python
grouped_data = df.groupby('category')['price'].agg(['mean', 'sum', 'count'])

grouped_data
```

```python
# Bar chart for mean values
grouped_data['mean'].plot(kind='bar', color='skyblue', title='Mean Values by Category')
plt.xlabel('Category')
plt.ylabel('Mean Value')
plt.show()

# Pie chart for sum values
grouped_data['sum'].plot(kind='pie', autopct='%1.1f%%', title='Sum Values by Category')
plt.ylabel('')
plt.show()
```
